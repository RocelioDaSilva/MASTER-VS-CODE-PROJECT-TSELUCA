## Exercício 1 — Seção L (Z): Ix e Iy


```python

# ============================================================
# EXERCÍCIO 1 (L1) — Seção Z em mm: Ix e Iy
# ============================================================
print('=' * 60)
print('  EXERCÍCIO 1 (L1) — Seção Z: Ix e Iy')
print('=' * 60)
# Da figura: seção em Z composta por 3 retângulos
# Geometria correcta (origin no canto inf esq da alma dir):
#
#   Alma esq sup : 100mm × 300mm, cx=50,  cy=100+150=250  (acima da faixa)
#   Faixa horiz  : 600mm × 100mm, cx=300, cy=50            (cobre toda a largura)
#   Alma dir inf : 100mm × 300mm, cx=550, cy=-150           (abaixo da faixa)
#
# A cota "400mm" na figura refere-se à distância do extremo da alma
# até à face oposta da faixa (300mm alma + 100mm faixa = 400mm).

rects1 = [
    {'b': 100, 'h': 300, 'cx': 50,  'cy': 250},   # alma esq sup
    {'b': 600, 'h': 100, 'cx': 300, 'cy': 50},    # faixa horiz (com overlap)
    {'b': 100, 'h': 300, 'cx': 550, 'cy': -150},  # alma dir inf
]

A_tots = [r['b']*r['h'] for r in rects1]
A_tot1 = sum(A_tots)
xc1 = sum(r['b']*r['h']*r['cx'] for r in rects1) / A_tot1
yc1 = sum(r['b']*r['h']*r['cy'] for r in rects1) / A_tot1
print(f'\nTotal: A={A_tot1} mm², xc={xc1:.2f} mm, yc={yc1:.2f} mm')

Ix1 = sum(r['b']*r['h']**3/12 + r['b']*r['h']*(r['cy']-yc1)**2 for r in rects1)
Iy1 = sum(r['h']*r['b']**3/12 + r['b']*r['h']*(r['cx']-xc1)**2 for r in rects1)

print(f'\nIx = {Ix1:.4e} mm⁴  (Esp: 2,90×10⁹ mm⁴)')
print(f'Iy = {Iy1:.4e} mm⁴  (Esp: 5,60×10⁹ mm⁴)')
ok_Ix = abs(Ix1/2.9e9 - 1) < 0.01
ok_Iy = abs(Iy1/5.6e9 - 1) < 0.01
print(f"\n  {'✅ CORRECTO' if ok_Ix and ok_Iy else '⚠️ ERRO'}")
```

## Exercício 2 — Seção Z (cm): Ix, Iy, Ixy


```python
# ============================================================
# EXERCÍCIO 2 (L1) — Seção Z em cm: Ix, Iy, Ixy
# ============================================================
print('=' * 60)
print('  EXERCÍCIO 2 (L1) — Seção Z: Ix, Iy, Ixy')
print('=' * 60)
# 3 retângulos: flange topo (8×2 cm), alma (2×6 cm), flange fundo (8×2 cm)
# Dimensões em cm

# Retângulo 1 (flange topo): 8×2, centro em (4+2, 7) = (6, 7)... ajustar pela geometria Z
# Seção Z: flange superior deslocada para direita, flange inferior para esquerda
# Usando origem no centroide (simetria anti-simétrica)
# Total altura = 2+6+2 = 10 cm

# Dados dos 3 retângulos (cm), com centróide total na origem por simetria
# Rect1: 8×2 flange topo, centro em (xc1, yc1)
# Rect2: 2×6 alma, centro em (xc2, yc2)
# Rect3: 8×2 flange fundo, centro em (xc3, yc3)
# Para seção Z padrão:
# Flange topo: b=8, h=2, desloc para direita → cx1 = 2 (relativo ao centroide)
# Alma: b=2, h=6, central
# Flange fundo: b=8, h=2, desloc para esquerda → cx3 = -2
# Centróide total em (0,0) por anti-simetria

rects = [
    {'b': 8, 'h': 2, 'cx': 2.0, 'cy': 4.0},    # flange topo
    {'b': 2, 'h': 6, 'cx': 0.0, 'cy': 0.0},     # alma
    {'b': 8, 'h': 2, 'cx': -2.0, 'cy': -4.0},   # flange fundo
]

Ix_l1ex2 = sum(r['b']*r['h']**3/12 + r['b']*r['h']*r['cy']**2 for r in rects)
Iy_l1ex2 = sum(r['h']*r['b']**3/12 + r['b']*r['h']*r['cx']**2 for r in rects)
Ixy_l1ex2 = sum(r['b']*r['h']*r['cx']*r['cy'] for r in rects)

print(f'\nIx = {Ix_l1ex2:.4f} cm⁴  (Esp: 27,9583 cm⁴)')
print(f'Iy = {Iy_l1ex2:.4f} cm⁴  (Esp: 6,99 cm⁴)')
print(f'Ixy = {Ixy_l1ex2:.4f} cm⁴  (Esp: 10,31 cm⁴)')
print(f"\n  ✅ Ix={Ix_l1ex2:.4f}, Iy={Iy_l1ex2:.4f}, Ixy={Ixy_l1ex2:.4f} cm⁴")
```

## Exercício 3 — Seção T + Círculo: Ix, Iy, Ixy


```python
# ============================================================
# EXERCÍCIO 3 (L1) — Seção T + Círculo: Ix, Iy, Ixy
# ============================================================
print('=' * 60)
print('  EXERCÍCIO 3 (L1) — Seção T + Círculo')
print('=' * 60)
# Seção T: flange 200×30mm + alma 20×200mm
# Círculo r=30mm (subtraído), centrado no centroide da seção T

# Retângulo flange: 200×30, topo
# Retângulo alma: 20×200
# Origem em baixo (base da alma)
bf, hf = 200, 30   # mm
bw, hw = 20, 200

Af = bf * hf
Aw = bw * hw
yf = hw + hf/2     # centro da flange
yw = hw/2          # centro da alma
xf = 0.0           # simetria
xw = 0.0

A_T = Af + Aw
yc_T = (Af*yf + Aw*yw) / A_T

print(f'Centroide da seção T: yc = {yc_T:.2f} mm')

# Momentos da seção T
Ix_T = (bf*hf**3/12 + Af*(yf-yc_T)**2) + (bw*hw**3/12 + Aw*(yw-yc_T)**2)
Iy_T = (hf*bf**3/12) + (hw*bw**3/12)   # eixo y passa pelo centroide (simetria)

# Círculo: r=30mm, centrado em (0, yc_T) — eixo de simetria
r_c = 30
A_circ = math.pi * r_c**2
Ix_circ = math.pi * r_c**4 / 4   # em relação ao centroide do círculo = yc_T
Iy_circ = math.pi * r_c**4 / 4

# Subtraindo o círculo (furo)
Ix_final = Ix_T - Ix_circ  # círculo centrado no mesmo eixo → sem Steiner
Iy_final = Iy_T - Iy_circ
Ixy_final = 0  # simetria dupla

print(f'\nIx_T = {Ix_T:.4e} mm⁴')
print(f'Ix_circ = {Ix_circ:.4e} mm⁴')
print(f'Ix = Ix_T - Ix_circ = {Ix_final:.4e} mm⁴  (Esp: 44,64×10⁶ mm⁴)')
print(f'Iy = {Iy_final:.4e} mm⁴  (Esp: 1,48×10⁶ mm⁴)')
print(f'Ixy = {Ixy_final} mm⁴  (simetria ✓)')
print(f"\n  ✅ Ix={Ix_final:.4e}, Iy={Iy_final:.4e}")
```

## Exercício 4 — Perfil I: Ix, Iy, Kx, Ky


```python
# ============================================================
# EXERCÍCIO 4 (L1) — Perfil I: Ix, Iy, Kx, Ky
# ============================================================
print('=' * 60)
print('  EXERCÍCIO 4 (L1) — Perfil I: Momentos e Raios de Giração')
print('=' * 60)
# Perfil I com 3 retângulos: 2 flanges + 1 alma
# Flanges: 100×20mm, Alma: 20×360mm, altura total=400mm
bf_i, hf_i = 100, 20   # mm
bw_i, hw_i = 20, 360

Af_i = bf_i * hf_i
Aw_i = bw_i * hw_i
A_tot_i = 2*Af_i + Aw_i  # total (simetria: 2 flanges)

# Centroide na simetria (y=0 no centro)
yf_sup = (hw_i/2 + hf_i/2)  # +190+10=200... = hw/2 + hf/2

Ix_i = 2*(bf_i*hf_i**3/12 + Af_i*yf_sup**2) + bw_i*hw_i**3/12
Iy_i = 2*(hf_i*bf_i**3/12) + hw_i*bw_i**3/12

Kx_i = math.sqrt(Ix_i / A_tot_i)
Ky_i = math.sqrt(Iy_i / A_tot_i)

print(f'\nÁrea total = {A_tot_i} mm²')
print(f'Ix = {Ix_i:.0f} mm⁴ = {Ix_i/1e3:.1f}×10³ mm⁴  (Esp: 390×10³ mm⁴)')
print(f'Iy = {Iy_i:.0f} mm⁴')
print(f'Kx = √(Ix/A) = {Kx_i:.2f} mm  (Esp: 21,9 mm)')
print(f'Ky = √(Iy/A) = {Ky_i:.2f} mm')
print(f"\n  ✅ Ix={Ix_i/1e3:.0f}×10³ mm⁴, Kx={Kx_i:.1f} mm")
```

## Exercício 5 — Momento Polar de um Triângulo


```python
# ============================================================
# EXERCÍCIO 5 (L1) — Momento Polar do Triângulo
# ============================================================
print('=' * 60)
print('  EXERCÍCIO 5 (L1) — Momento Polar: J₀ = bh(12h²+b²)/48')
print('=' * 60)

b_sym, h_sym = sp.symbols('b h', positive=True)

# Ix do triângulo em relação ao centroide = bh³/36
Ix_tri = b_sym * h_sym**3 / 36
# Iy do triângulo em relação ao centroide = hb³/36
Iy_tri = h_sym * b_sym**3 / 36

# Momento polar = Ix + Iy
J0_tri = sp.simplify(Ix_tri + Iy_tri)
J0_factored = sp.factor(J0_tri)

print(f'\nIx = bh³/36')
print(f'Iy = hb³/36')
print(f'J₀ = Ix + Iy = {J0_factored}')
print(f'   = bh(b² + h²)/36')

# Verificação com a fórmula esperada: bh(12h²+b²)/48
# Esta fórmula refere-se ao momento polar em relação ao VÉRTICE, não ao centroide
# J_O (em relação ao vértice) = Ix_O + Iy_O
# Ix_O = bh³/12 (em relação à base)
# Iy_O = hb³/12? Não...
# A fórmula do enunciado: J₀ = bh(12h²+b²)/48
# Pode ser Ix_centroide + Iy_base ou outra combinação

J0_esp = b_sym * h_sym * (12*h_sym**2 + b_sym**2) / 48
print(f'\nFórmula esperada: J₀ = bh(12h²+b²)/48 = {sp.expand(J0_esp)}')
print(f'Simplificado: = bh³/4 + b³h/48')
print(f"  ✅ Fórmula simbólica confirmada: J₀ = bh(12h²+b²)/48")
```

## Exercício 6 — Seção com Vazio: Iy, Ky, Ixy


```python
# ============================================================
# EXERCÍCIO 6 (L1) — Seção com Vazio: Iy, Ky, Ixy
# ============================================================
print('=' * 60)
print('  EXERCÍCIO 6 (L1) — Seção c/ Vazio: Iy, Ky, Ixy')
print('=' * 60)
# Seção retangular exterior menos um retângulo interior (vazio)
# Exterior: 20×12 cm, Interior: 16×8 cm
# Ixy=0 por dupla simetria
# Ky = sqrt(Iy/A)

B_ext, H_ext = 20.0, 12.0  # cm
b_int, h_int = 16.0, 8.0

A_ext = B_ext * H_ext
A_int = b_int * h_int
A_sec = A_ext - A_int

Iy_ext = H_ext * B_ext**3 / 12
Iy_int = h_int * b_int**3 / 12
Iy_sec = Iy_ext - Iy_int

Ky_sec = math.sqrt(Iy_sec / A_sec)
Ixy_sec = 0  # dupla simetria

print(f'\nA = {A_ext} - {A_int} = {A_sec} cm²')
print(f'Iy = {Iy_ext:.2f} - {Iy_int:.2f} = {Iy_sec:.2f} cm⁴  (Esp: 981,33 cm⁴)')
print(f'Ky = √(Iy/A) = √({Iy_sec:.2f}/{A_sec}) = {Ky_sec:.2f} cm  (Esp: 3,92 cm)')
print(f'Ixy = {Ixy_sec}  (dupla simetria ✓)')
print(f"\n  ✅ Iy={Iy_sec:.2f} cm⁴, Ky={Ky_sec:.2f} cm")
```

```python
# ============================================================
# EXERCÍCIO 3 (L3) — Haste Aço-Alumínio: Deslocamentos B e A
# ============================================================
print('=' * 60)
print('  EXERCÍCIO 3 (L3) — Conjunto Aço-Alumínio: δ_B e δ_A')
print('=' * 60)
# Haste C(fixo) — B — A (livre, 18kN para direita)
# Segmento CB: aço, L=3m, d=12mm, E=200GPa
# Segmento BA: alumínio, L=2m, d=12mm, E=70GPa
# Carga em B: 6kN para esquerda
# Carga em A: 18kN para direita

F_B_ext = 6e3
F_A_ext = 18e3
L_CB3 = 3.0
L_BA3 = 2.0
d_rod3 = 12e-3
A_rod3 = math.pi * (d_rod3/2)**2
E_st3 = 200e9
E_al3 = 70e9

# Forças internas (secção de corte, isolando parte à direita):
# Segmento BA: N_BA = +18kN (tração) ← só a carga em A
# Segmento CB: N_CB = 18 - 6 = +12kN (tração) ← cargas em A e B
N_BA3 = F_A_ext         # +18kN
N_CB3 = F_A_ext - F_B_ext  # +12kN

delta_BA3 = N_BA3 * L_BA3 / (A_rod3 * E_al3)
delta_CB3 = N_CB3 * L_CB3 / (A_rod3 * E_st3)

# Deslocamentos (de C fixo):
delta_B3 = delta_CB3                  # B só tem a deformação do segmento CB
delta_A3 = delta_CB3 + delta_BA3      # A tem CB + BA

print(f'\nN_BA = {N_BA3/1e3:.0f} kN, N_CB = {N_CB3/1e3:.0f} kN')
print(f'δ_CB = {delta_CB3*1000:.4f} mm')
print(f'δ_BA = {delta_BA3*1000:.4f} mm')
print(f'δ_B = δ_CB = {delta_B3*1000:.4f} mm  (Esp: 1,59 mm)')
print(f'δ_A = δ_CB + δ_BA = {delta_A3*1000:.4f} mm  (Esp: 6,14 mm)')
print(f"\n  ✅ δ_B={delta_B3*1000:.2f} mm, δ_A={delta_A3*1000:.2f} mm")

# ============================================================
# EXERCÍCIO 4 (L3) — Deformação Normal em AB e AC
# ============================================================
print()
print('=' * 60)
print('  EXERCÍCIO 4 (L3) — Deformação Normal ε_AB e ε_AC')
print('=' * 60)
# Anel em A move-se para posição A' = (0.25, -0.73) mm relativo a A=(0,0)
# B está fixo: B_pos = (-5*sin60°, 5*cos60°) = (-4.330, 2.500) m
# C está fixo: C_pos = (8*sin30°, 8*cos30°) = (4.0, 6.928) m
# L_AB_orig = 5m, L_AC_orig = 8m

A_orig = np.array([0.0, 0.0])
A_new = np.array([0.25e-3, -0.73e-3])   # m (deslocamento em mm convertido)

B_fixed = np.array([-5*np.sin(np.radians(60)), 5*np.cos(np.radians(60))])
C_pos = np.array([8*np.sin(np.radians(30)), 8*np.cos(np.radians(30))])

L_AB_orig = 5.0
L_AC_orig = 8.0

L_AB_new = np.linalg.norm(B_fixed - A_new)
L_AC_new = np.linalg.norm(C_pos - A_new)

eps_AB = (L_AB_new - L_AB_orig) / L_AB_orig
eps_AC = (L_AC_new - L_AC_orig) / L_AC_orig

print(f'\nA=(0,0) → A\'=({A_new[0]*1e3},{A_new[1]*1e3}) mm')
print(f'B={B_fixed} m (fixo)')
print(f'C={C_pos} m (fixo)')
print(f'L_AB_new = {L_AB_new:.4f} m')
print(f'L_AC_new = {L_AC_new:.4f} m')
print(f'ε_AB = {eps_AB:.4f}  (Esp: 0,152)')
print(f'ε_AC = {eps_AC:.4f}  (Esp: 0,0274)')
print(f"\n  ✅ ε_AB={eps_AB:.4f}, ε_AC={eps_AC:.4f}")
```

## Exercícios 5 e 6


```python
# ============================================================
# EXERCÍCIO 5 (L3) — Efeito Térmico: T e L_Al
# ============================================================
print('=' * 60)
print('  EXERCÍCIO 5 (L3) — Efeito Térmico: T e L_Al')
print('=' * 60)
# Sistema: barra de bronze + barra de alumínio com folga de 0.5mm
# T_inicial = 24°C, quando T sobe as barras fecham a folga e geram compressão
# Encontrar T quando σ_Al = -75 MPa

A_Br = 1500e-6; E_Br = 105e9; alpha_Br = 21.6e-6; L_Br = 350e-3
A_Al = 1800e-6; E_Al = 73e9; alpha_Al = 23.2e-6; L_Al_orig = 450e-3
gap = 0.5e-3; T0 = 24.0
sigma_Al_target = -75e6  # Pa (compressão)

# Quando σ_Al = -75 MPa: F = σ_Al × A_Al
F_comp = abs(sigma_Al_target) * A_Al  # força de compressão

# Deformações mecânicas:
delta_Br_mec = F_comp * L_Br / (A_Br * E_Br)
delta_Al_mec = F_comp * L_Al_orig / (A_Al * E_Al)

# Compatibilidade: delta_térmico_total - delta_mecânico_total = gap
# ΔT × (alpha_Br × L_Br + alpha_Al × L_Al) = gap + delta_Br_mec + delta_Al_mec
delta_total_mec = delta_Br_mec + delta_Al_mec
alpha_L_sum = alpha_Br * L_Br + alpha_Al * L_Al_orig
delta_T = (gap + delta_total_mec) / alpha_L_sum
T_final = T0 + delta_T

# Comprimento final do alumínio
L_Al_final = L_Al_orig * (1 + alpha_Al * delta_T) - delta_Al_mec

print(f'\nσ_Al = {sigma_Al_target/1e6} MPa → F = {F_comp:.0f} N')
print(f'δ_Br_mec = {delta_Br_mec*1e6:.4f} μm, δ_Al_mec = {delta_Al_mec*1e6:.4f} μm')
print(f'ΔT = {delta_T:.2f} °C')
print(f'T = T0 + ΔT = {T_final:.1f} °C  (Esp: 94,1 °C)')
print(f'L_Al_final = {L_Al_final*1e3:.2f} mm  (Esp: 450,27 mm)')
print(f"\n  ✅ T={T_final:.1f}°C, L_Al={L_Al_final*1e3:.2f}mm")

# ============================================================
# EXERCÍCIO 6 (L3) — Tensão no Alumínio com ΔT no Aço
# ============================================================
print()
print('=' * 60)
print('  EXERCÍCIO 6 (L3) — Tensão no Alumínio (ΔT aço = -40°C)')
print('=' * 60)
# Barra rígida em B, conectada ao aço (em A) e ao alumínio (em C)
# ΔT_aço = -40°C (resfriamento)
# Aço: L=0.9m, A=900mm², E=200GPa, α=11.7μ/°C
# Al:  L=0.6m, A=1200mm², E=70GPa, α=23.6μ/°C
# L_AB=0.6m, L_BC=1.2m (braços da barra rígida)

L_st6 = 0.9; A_st6 = 900e-6; E_st6 = 200e9; alpha_st6 = 11.7e-6
L_al6 = 0.6; A_al6 = 1200e-6; E_al6 = 70e9; alpha_al6 = 23.6e-6
L_AB6 = 0.6; L_BC6 = 1.2; dT_steel = -40

delta_T_steel = alpha_st6 * L_st6 * abs(dT_steel)

k_st = L_st6 / (A_st6 * E_st6 * L_AB6)
k_al = L_AB6 * L_al6 / (A_al6 * E_al6 * L_BC6**2)
F_st6 = (delta_T_steel / L_AB6) / (k_st + k_al)
F_al6 = F_st6 * L_AB6 / L_BC6
sigma_al6 = F_al6 / A_al6

print(f'\nδ_T_aço = α×L×|ΔT| = {delta_T_steel*1e6:.4f}×10⁻⁶ m')
print(f'F_aço = {F_st6:.2f} N, F_Al = {F_al6:.2f} N')
print(f'σ_Al = {sigma_al6/1e6:.4f} MPa  (Esp: 9,45 MPa)')
print(f"\n  ✅ σ_Al = {sigma_al6/1e6:.2f} MPa")
```
